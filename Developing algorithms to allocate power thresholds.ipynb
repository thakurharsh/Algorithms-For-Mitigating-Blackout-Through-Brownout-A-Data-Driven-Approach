```python
def cu(data):
    cud = pd.DataFrame()
    cud = data
    for x in range(0,23):
        cud['max_h_'+str(x)]=cud['h_'+str(x)].max()
    for x in range(0,23):
        cud['ratio_h_init_'+str(x)]=(cud['h_'+str(x)]/cud['max_h_'+str(x)])
    for x in range(0,23):
        cud.loc[cud['ratio_h_init_'+str(x)] < 0 , 'ratio_h_init_'+str(x)] = 0
    for x in range(0,23):
        cud['ratio_h_'+str(x)]=(1-cud['ratio_h_init_'+str(x)])
    cud['den']=cud['ratio_h_0']+cud['ratio_h_1']+cud['ratio_h_2']+cud['ratio_h_3']+cud['ratio_h_4']+cud['ratio_h_5']+cud['ratio_h_6']+cud['ratio_h_7']+cud['ratio_h_8']+cud['ratio_h_9']+cud['ratio_h_10']+cud['ratio_h_11']+cud['ratio_h_12']+cud['ratio_h_13']+cud['ratio_h_14']+cud['ratio_h_15']+cud['ratio_h_16']+cud['ratio_h_17']+cud['ratio_h_18']+cud['ratio_h_19']+cud['ratio_h_20']+cud['ratio_h_21']+cud['ratio_h_22']
    for x in range(0,23):
        cud['CUD_h_'+str(x)]=((cud['ratio_h_'+str(x)]/cud['den'])*cud['Big Delta'])
    for x in range(0,23):
        cud.loc[cud['CUD_h_'+str(x)] < 0 , 'CUD_h_'+str(x)] = 0
    for x in range(0,23):
        cud['CUD_VI_h_'+str(x)]=(cud['h_'+str(x)]-cud['CUD_h_'+str(x)])/cud['h_'+str(x)]
    for x in range(0,23):
        cud.loc[cud['CUD_VI_h_'+str(x)] < 0 , 'CUD_VI_h_'+str(x)] = 0
    cud=cud.fillna(0)
    it =[item for item in range(0, 23)]
    #figure for violation
    fig,axs = plt.subplots(nrows=5,ncols=5,figsize = (25,12))
    plt.subplots_adjust(hspace=0.75)
    fig.suptitle('Consumer Usage distribution for California Violation Ratio house wise')
    for x, ax in zip(it, axs.ravel()):
        cud.plot(x='localminute',y='CUD_VI_h_'+str(x),xlabel='calendar day',ylabel='Violation_Ratio',ax=ax )
        ax.set_title('House_'+str(x))
        ax.get_legend().remove()
        ax.set_xlabel("calendar day")
        
    plt.show()

    #figure for comparison between comsumption and new threshold
    fig,axs = plt.subplots(nrows=5,ncols=5,figsize = (25,12))
    plt.subplots_adjust(hspace=0.75)
    fig.suptitle('Power consumption vs small delta i.e Consumer Usage distribution delta here for California house wise')
    for x, ax in zip(it, axs.ravel()):
        cud.plot(x='localminute',y=['h_'+str(x),'CUD_h_'+str(x)],xlabel='calendar day',ylabel='kilo watt',ax=ax )
        ax.set_title('House_'+str(x))
        #x.get_legend().remove()
        ax.set_xlabel("calendar day")
        ax.set_ylim(0, 3)
        ax.legend(["power consumption", "Consumer usage distribution"], loc ="best")
    plt.show()
    
    cud['CUD_cal_ND'] = (cud['CUD_VI_h_0']+cud['CUD_VI_h_1']+cud['CUD_VI_h_2']+cud['CUD_VI_h_3']+cud['CUD_VI_h_4']+cud['CUD_VI_h_5']+cud['CUD_VI_h_6']+cud['CUD_VI_h_7']+cud['CUD_VI_h_8']+cud['CUD_VI_h_9']+cud['CUD_VI_h_10']+cud['CUD_VI_h_11']+cud['CUD_VI_h_12']+cud['CUD_VI_h_13']+cud['CUD_VI_h_14']+cud['CUD_VI_h_15']+cud['CUD_VI_h_16']+cud['CUD_VI_h_17']+cud['CUD_VI_h_18']+cud['CUD_VI_h_19']+cud['CUD_VI_h_20']+cud['CUD_VI_h_21']+cud['CUD_VI_h_22'])/23
    # df_cal=pd.DataFrame()
    #df_cal['CUD_cal_ND']=cud['CUD_cal_ND']
    cud.plot(y='CUD_cal_ND',x='localminute',xlabel='calendar day',ylabel='Violation_Ratio',title='Consumer Usage distribution violation for California neighborhood',use_index=True)
    plt.ylim(0,1)
    plt.show()
    
    box_plot_data=[cud['CUD_VI_h_0'],cud['CUD_VI_h_1'],cud['CUD_VI_h_2'],cud['CUD_VI_h_3'],cud['CUD_VI_h_4'],cud['CUD_VI_h_5'],cud['CUD_VI_h_6'],cud['CUD_VI_h_7'],cud['CUD_VI_h_8'],cud['CUD_VI_h_9'],cud['CUD_VI_h_10'],cud['CUD_VI_h_11'],cud['CUD_VI_h_12'],cud['CUD_VI_h_13'],cud['CUD_VI_h_14'],cud['CUD_VI_h_15'],cud['CUD_VI_h_16'],cud['CUD_VI_h_17'],cud['CUD_VI_h_18'],cud['CUD_VI_h_19'],cud['CUD_VI_h_20'],cud['CUD_VI_h_21'],cud['CUD_VI_h_22']]
    plt.title("California Consumer Usage distribution")
    plt.boxplot(box_plot_data,labels=['h1','h2','h3','h4','h5','h6','h7','h8','h9','h10','h11','h12','h13','h14','h15','h16','h17','h18','h19','h20','h21','h22','h23'])
    plt.xlabel('houses')
    plt.ylabel('%violation')
    plt.show()
    global df_cu_cal
    df_cu_cal = pd.DataFrame(cud)
    return df_cu_cal

    
    
```
